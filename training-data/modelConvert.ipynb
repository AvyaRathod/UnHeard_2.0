```python
import os
import coremltools
from keras.models import load_model

def main():
	model_name = "model_checkpoint.h5"
	keras_model = load_model(model_name, compile =False)

	mlmodel = coremltools.convert(keras_model, convert_to="mlprogram", source="tensorflow")


	f_name, f_ext = os.path.splitext(model_name)
	mlmodel.save(os.path.join(f_name + '.mlpackage'))

	print("Success!")

if __name__ == '__main__':
	main()	
```
